Duyệt theo chiều rộng (BFS) của trò chơi 8 ô
Ma trận 3x3:
1 2 3
4 5 6
8 0 7

Flow BFS:
1. Lấy node đầu frontier
2. Expand (xét)
3. Sinh node con
4. Kiểm tra reached
5. Thêm frontier
6. Lặp

# frontier chứa các node chưa EXPAND

vd: Node A có
    1 2 3 
    4 5 6 , _ , _ , _
    8 0 7

Node | Frontier                                | Reached
A    | [ 1,2,3 | 4,5,6 | 8,7,0], A,R,1 -> B    | [A]
     | [ 1,2,3 | 4,5,6 | 0,8,7], A,L,1 -> C    |
     | [ 1,2,3 | 4,0,6 | 8,5,7], A,U,1 -> D    |
     | [B, C, D]                               |
B    | [ 1,2,3 | 4,5,0 | 8,7,6], B,U,2 -> E    | [A, B]
     | [C, D, E]                               |
....

```python
from collections import deque
```

```python
## START & GOAL 

s = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 0, 8]
]

g = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]
```

```python
## HÀM IN MA TRẬN 

def print_state(state):

    for row in state:
        print(row)
    print()
```

```python
## TÌM VỊ TRÍ Ô 0 

def find_zero(state):

    for i in range(3):

        for j in range(3):

            if state[i][j] == 0:
                return i, j
```

```python
## SINH NODE CON 

def generate_children(state):

    children = []

    x, y = find_zero(state)

    moves = []

    # Up
    if x > 0:
        moves.append(("U", x - 1, y))

    # Down
    if x < 2:
        moves.append(("D", x + 1, y))

    # Left
    if y > 0:
        moves.append(("L", x, y - 1))

    # Right
    if y < 2:
        moves.append(("R", x, y + 1))

    for action, nx, ny in moves:

        # copy ma trận
        new_state = [row[:] for row in state]

        # swap
        new_state[x][y], new_state[nx][ny] = \
        new_state[nx][ny], new_state[x][y]

        children.append((action, new_state))
        
    return children
```

```python
#Chương trình chính
# BFS
frontier = deque()

reached = set()

# tên node
node_char = ord('A')

# node đầu
frontier.append((chr(node_char), s, 0))

step = 0

while frontier:

    print("\n========== STEP", step, "==========")

    # lấy node đầu
    node_name, current_state, path_cost = frontier.popleft()

    print("Current Node =", node_name)
    print("Path Cost =", path_cost)
    print("\nState:")
    print_state(current_state)

    state_tuple = tuple(map(tuple, current_state))

    # tránh lặp
    if state_tuple in reached:
        continue

    reached.add(state_tuple)

    # hiện reached
    print("Reached:")

    for r in reached:
        print(r)

    print()

    # kiểm tra goal
    if current_state == g:

        print("GOAL FOUND!")
        break

    # sinh node con
    children = generate_children(current_state)

    print("Children:")

    for action, child_state in children:

        child_tuple = tuple(map(tuple, child_state))

        if child_tuple not in reached:

            # tăng tên node
            node_char += 1

            child_name = chr(node_char)
            child_cost = path_cost + 1

            print("\n--------------------------------\n")

            print(node_name, ",", action,
                  ", cost =", child_cost,
                  "→", child_name)

            print_state(child_state)

            frontier.append(
                (child_name, child_state, child_cost)
            )

    # hiện frontier
    print("Frontier:")

    for f in frontier:
        print(f[0], end=" ")

    print()

    step += 1
```
