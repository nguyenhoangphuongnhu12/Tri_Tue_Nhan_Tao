Mô hình Simple Reflex Agent của trò chơi 8 ô

Environment
Ma trận 3x3:
1 2 3
4 6 5
8 _ 7

RULE
If state = (0,0) then active = RAND(D,R)
If state = (0,1) then active = RAND(D,R,L)
If state = (0,2) then active = RAND(D,L)
If state = (1,0) then active = RAND(U,D,R)
If state = (1,1) then active = RAND(U,D,R,L)
If state = (1,2) then active = RAND(U,D,L)
If state = (2,0) then active = RAND(U,R)
If state = (2,1) then active = RAND(U,R,L)
If state = (2,2) then active = RAND(U,L)

Chọn hành động
Active = RAND(Moves)

Ví dụ
Ô trống có state = (2,1)
Chiếu theo RULE:
If state = (0,1) then active = RAND(U,R,L)
=> Ô trống được di chuyển qua trái/phải, lên
Danh sách Moves = [U,R,L]

Điều kiện dừng = Thắng game
If Ma trận 3x3:
1 2 3
4 5 6
7 8 _

```python
import random
import time
```

```python
#Kích thước ma trận
m = 3
n = 3

#Ô trống: []

#Ma trận 3x3
my_board = [
    [1,2,3],
    [4,5,6],
    [7,[],8]
]

#Vị trí hiện tại của ô trống []
x = 2
y = 1
```

```python
#Kiểm tra điều kiện dừng/thắng game
def check_win(current_board):
    win_board = [
         [1, 2, 3],
        [4, 5, 6],
        [7, 8, []]
    ]
    
    #So sánh trực tiếp 2 ma trận
    return current_board == win_board

if check_win(my_board):
    print("Bạn đã thắng!")
else:
    print("Chưa thắng, chơi tiếp đi!")
```

Hàm in ma trận
1 2 3
4 6 5
8 _ 7

```python
def print_board(my_board):
    for i in range(m):
        for j in range(n):
            val = my_board[i][j]
            
            #Kiểm tra nếu là ô trống [] thì in khoảng trắng, ngược lại in số
            if val == []:
                print(" ", end = " ")
            else:
                print(val, end = " ")
                
        print() #Xuống dòng sau khi in xong một hàng
```

Chương trình chính
Bước 1: Kiểm tra state của ô trống []
Bước 2: Dựa trên RULE, tìm hướng đi hợp lệ
Bước 3: Random hướng đi
Bước 4: Cập nhật vị trí ô trống []

```python
step = 1

#Chạy liên tục khi chưa thắng game
while not check_win(my_board):
    print("\n===== STEP", step, "=====")
    print_board(my_board)
    
    #1.Tìm vị trí hiện tại của ô trống []
    empty_i, empty_j = -1, -1
    for i in range(m):
        for j in range(n):
            if my_board[i][j] == []:
                empty_i, empty_j = i, j
                break
        if empty_i != -1:  # SỬA LỖI: Thoát hẳn vòng lặp ngoài khi đã tìm thấy ô trống
            break
        
    #BƯỚC 2: Dựa trên RULE, tìm hướng đi hợp lệ (Không vượt quá biên 3x3)
    #Định nghĩa quy tắc dịch chuyển ô trống: (Tên hướng, delta_i, delta_j)
    rules = {
        'UP': (-1, 0),
        'DOWN': (1, 0),
        'LEFT': (0, -1),
        'RIGHT': (0, 1)
    }
    
    moves = [] # Danh sách lưu các hướng đi hợp lệ tại vị trí hiện tại
    
    for direction, (di, dj) in rules.items():
        new_i = empty_i + di
        new_j = empty_j + dj
        # Nếu tọa độ mới nằm trong phạm vi ma trận 3x3
        if 0 <= new_i < 3 and 0 <= new_j < 3:
            moves.append((direction, new_i, new_j))

    # BƯỚC 3: Random hướng đi từ các hướng hợp lệ tìm được
    chosen_move = random.choice(moves)
    direction_name, target_i, target_j = chosen_move
    print(f"Agent phản xạ chọn hướng ngẫu nhiên: {direction_name}")

    # BƯỚC 4: Cập nhật vị trí ô trống [] (Hoán đổi giá trị số và [])
    my_board[empty_i][empty_j] = my_board[target_i][target_j]
    my_board[target_i][target_j] = []
    
    # Tăng bước và tạo độ trễ 0.5 giây để dễ quan sát màn hình chạy tự động
    step += 1
    time.sleep(0.5)

# Khi thoát vòng lặp đồng nghĩa Agent đã thắng game
print("\n===== TRẠNG THÁI CUỐI CÙNG =====")
print_board(my_board)
print(f"Agent đã đưa trò chơi về trạng thái đích sau {step - 1} bước!")
```
