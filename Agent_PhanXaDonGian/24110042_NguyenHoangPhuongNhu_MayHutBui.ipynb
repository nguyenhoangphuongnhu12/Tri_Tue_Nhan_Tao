# Mô hình Simple Reflex Agent của Máy hút bụi

## Environment

Ma trận 3x3:

- 0 = ô sạch
- 1 = ô có bụi
- 2 = robot

# RULE 1 — Hút bụi

IF State(x,y) = 1 THEN Action = SUCK

và cập nhật State(x,y) = 0

# RULE 2 — Di chuyển

IF State(x,y) = 0 THEN Robot sẽ tìm các hướng hợp lệ.

## Kiểm tra đi lên

IF x > 0 THEN Move.append(U)

Ý nghĩa:
- robot chưa chạm biên trên
- nên được đi lên

## Kiểm tra đi xuống

IF x < m - 1 THEN Move.append(D)

Ý nghĩa:
- robot chưa chạm biên dưới
- nên được đi xuống

## Kiểm tra đi trái

IF y > 0 THEN Move.append(L)

Ý nghĩa:
- robot chưa chạm biên trái
- nên được đi trái

## Kiểm tra đi phải

IF y < n - 1 THEN Move.append(R)

Ý nghĩa:
- robot chưa chạm biên phải
- nên được đi phải

# Chọn hành động

Action = RAND(Move)

Robot sẽ random một hướng trong danh sách Move.

# Ví dụ

Robot ở State = (0,0)

THEN:

- không được đi lên
- không được đi trái

Danh sách Move = [D,R]

# Điều kiện dừng

IF Tất cả ô = 0 THEN STOP

```python
import random
import time
```

```python
# Kích thước phòng
m = 3
n = 3

# Ma trận phòng
room = [
    [1, 0, 1],
    [0, 1, 1],
    [1, 1, 0]
]

# Vị trí ban đầu của robot
x = 1
y = 1
```

```python
# Kiểm tra còn bụi không

def has_dust():

    for row in room:

        if 1 in row:

            return True

    return False
```

# Hàm in ma trận

- 2 = robot
- 1 = bụi
- 0 = sạch

```python
def print_room():

    for i in range(m):

        for j in range(n):

            if i == x and j == y:
                print("2", end=" ")

            else:
                print(room[i][j], end=" ")

        print()

    print()
```

# Chương trình chính

Bước 1:
- kiểm tra state hiện tại

Bước 2:
- nếu có bụi → hút bụi

Bước 3:
- nếu sạch → tìm các hướng hợp lệ

Bước 4:
- random hướng đi

Bước 5:
- cập nhật vị trí robot

```python
step = 1

# Chạy tới khi hết bụi

while has_dust():

    print("\n===== STEP", step, "=====")

    # RULE 1: Nếu có bụi

    if room[x][y] == 1:

        print("State =", room[x][y])

        print("Action = SUCK")

        room[x][y] = 0

    # RULE 2: Nếu sạch

    else:

        print("State =", room[x][y])

        moves = []

        # Kiểm tra đi lên
        if x > 0:

            moves.append("U")

        # Kiểm tra đi xuống
        if x < m - 1:

            moves.append("D")

        # Kiểm tra đi trái
        if y > 0:

            moves.append("L")

        # Kiểm tra đi phải
        if y < n - 1:

            moves.append("R")

        print("Possible Move =", moves)

        # Random hướng
        action = random.choice(moves)

        print("Action =", action)

        # Cập nhật vị trí

        if action == "U":

            x -= 1

        elif action == "D":

            x += 1

        elif action == "L":

            y -= 1

        elif action == "R":

            y += 1

    print_room()

    step += 1

    time.sleep(1)

print("ALL CLEAN !!!")
```
