Environment: Ma trận 3x3
Trạng thái hiện tại
1 2 3
4 5 6
8 _ 7

Trạng thái trước đó (ví dụ)
1 2 3
4 5 6
_ 8 7

B1: RULE
Kiểm tra biên 
    - Kiểm tra đi lên
    If x > 0 THEN move.append(U)
    - Kiểm tra đi xuống
    If x < m - 1 THEN move.append(D)
    - Kiểm tra qua trái
    If y > 0 THEN move.append(R)
    - Kiểm tra qua phải 
    If y < n - 1 THEN move.append(L)


Điều kiện dừng:
If ma trận 3x3:
1 2 3
4 5 6
7 8 _

```python
import random
import time
```

```python
#Kích thước ma trận
m = 3
n = 3

#Ô trống: []

#Ma trận 3x3
state = [
    [1,2,3],
    [4,5,6],
    [8,[],7]
]

#Vị trí hiện tại của ô trống []
x = 2
y = 1
```

```python
#Kiểm tra điều kiện dừng/thắng game
def check_win(current_state):
    win_board = [
         [1, 2, 3],
        [4, 5, 6],
        [7, 8, []]
    ]
    
    #So sánh trực tiếp 2 ma trận
    return current_state == win_board

if check_win(state):
    print("Bạn đã thắng!")
else:
    print("Chưa thắng, chơi tiếp đi!")
```

```python
#Hàm in ma trận
def print_board(state):
    for i in range(m):
        for j in range(n):
            val = state[i][j]
            
            #Kiểm tra nếu là ô trống [] thì in khoảng trắng, ngược lại in số
            if val == []:
                print(" ", end = " ")
            else:
                print(val, end = " ")
                
        print() #Xuống dòng sau khi in xong một hàng
```

    Chương trình chính

    Bước 1: Kiểm tra vị trí ô trống []
    tìm vị trí của ô trống

    Bước 2: Dựa trên RULE tìm hướng đi hợp lệ

    Bước 3: Random hướng đi
    action = random(directions)

    Bước 4: Cập nhật vị trí ô trống
    thực hiện action
    đổi vị trí ô trống với ô kế bên

    Bước 5: Cập nhật model/state
    UPDATE_MODEL(state, action, percept, model)

```python
# Model
model = {
    "state": state,
    "last_action": None
}

# UPDATE MODEL
def UPDATE_MODEL(state, action, percept, model):

    # 1. Nhận percept mới
    current_percept = percept

    # 2. Suy ra state mới
    new_state = current_percept

    # 3. Cập nhật model
    model["state"] = new_state
    model["last_action"] = action

    # 4. Lưu state mới
    state = new_state

    # 5. Trả về state mới
    return state
```

```python
step = 1
# Giới hạn số bước
MAX_STEP = 500

while not check_win(state) and step <= MAX_STEP:

    print("\n===== STEP", step, "=====")
    print_board(state)

    # 1. Tìm vị trí ô trống
    empty_i, empty_j = -1, -1

    for i in range(m):
        for j in range(n):

            if state[i][j] == 0:
                empty_i = i
                empty_j = j
                break

        if empty_i != -1:
            break

    # 2. Tìm hướng đi hợp lệ
    moves = []

    # Đi lên
    if empty_i > 0:
        moves.append("U")

    # Đi xuống
    if empty_i < m - 1:
        moves.append("D")

    # Đi trái
    if empty_j > 0:
        moves.append("L")

    # Đi phải
    if empty_j < n - 1:
        moves.append("R")

    print("Possible Move =", moves)

    # 3. Random hướng đi
    action = random.choice(moves)

    print("Action =", action)

    # percept = trạng thái agent quan sát được
    percept = [row[:] for row in state]

    # Di chuyển ô trống

    # UP
    if action == "U":

        percept[empty_i][empty_j], percept[empty_i - 1][empty_j] = \
        percept[empty_i - 1][empty_j], percept[empty_i][empty_j]

    # DOWN
    elif action == "D":

        percept[empty_i][empty_j], percept[empty_i + 1][empty_j] = \
        percept[empty_i + 1][empty_j], percept[empty_i][empty_j]

    # LEFT
    elif action == "L":

        percept[empty_i][empty_j], percept[empty_i][empty_j - 1] = \
        percept[empty_i][empty_j - 1], percept[empty_i][empty_j]

    # RIGHT
    elif action == "R":

        percept[empty_i][empty_j], percept[empty_i][empty_j + 1] = \
        percept[empty_i][empty_j + 1], percept[empty_i][empty_j]


    # UPDATE MODEL
    state = UPDATE_MODEL(state, action, percept, model)

    step += 1


print("LAST STATE")

print_board(state)
```
