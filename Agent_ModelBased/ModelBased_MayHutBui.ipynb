Mô hình Model Based Agent của máy hút bụi

Environment
Ma trận 3x3
0 1 1
1 1 0
0 0 1

0: sạch, 1: bẩn, 2: robot

Bước 1: Xây dựng lập luận (RULES)
RULE 1: Ô bẩn
    If current_state(x,y) = 1 THEN action = suck
                                AND current_state(x,y) = 0

RULE 2: Ô sạch
    If current_state(x,y) = 0 AND xét hướng đi U/D/R/L
                                THEN predict_state = vị trí mới

## Kiểm tra đi lên

IF x > 0 THEN move.append(U)

Ý nghĩa:
- robot chưa chạm biên trên
- nên được đi lên

## Kiểm tra đi xuống

IF x < m - 1 THEN move.append(D)

Ý nghĩa:
- robot chưa chạm biên dưới
- nên được đi xuống

## Kiểm tra đi trái

IF y > 0 THEN move.append(L)

Ý nghĩa:
- robot chưa chạm biên trái
- nên được đi trái

## Kiểm tra đi phải

IF y < n - 1 THEN move.append(R)

Ý nghĩa:
- robot chưa chạm biên phải
- nên được đi phải

# Chọn hành động

Action = RAND(Move)

Robot sẽ random một hướng trong danh sách Move.

RULE 3: Xét predict_state với current_state
    # Predict state
    predict_state = current_state

    If predict_state != current_state THEN action = move 
    Else:
        THEN action = stay

RULE 4: Điều kiện dừng
If all = 0  THEN stop

Ví dụ: current_state = (0,1)
        value_state = 1 -> bẩn -> hút bụi
        xét hướng có thể đi D/R/L -> predict_state = (0,2)
        predict_state != current_state -> move R -> current_state = predict_state 

```python
import random
import time
```

```python
# Kích thước phòng
m = 3
n = 3

# Ma trận phòng
room = [
    [0, 1, 1],
    [1, 1, 0],
    [0, 0, 1]
]

# Vị trí ban đầu của robot
x = 0
y = 0
```

```python
# Kiểm tra còn bẩn không
def has_dust():

    for i in range(m):

        for j in range(n):

            if room[i][j] == 1:
                return True

    return False
```

```python
#Hàm in ma trận
def print_room(x, y):

    for i in range(m):

        for j in range(n):

            if i == x and j == y:
                print("2", end=" ")

            else:
                print(room[i][j], end=" ")

        print()

    print()
```

```python
#Chương trình chính

step = 1

while has_dust():

    print("\n===== STEP", step, "=====")

    print_room(x, y)

    # #Bước 2: Tiếp nhận thông tin

    current_state = (x, y)

    value_state = room[x][y]

    print("Current State =", current_state)

    print("Value State =", value_state)


    # #Bước 3: Suy nghĩ

    # RULE 1:
    # IF current_state = 1
    # THEN Suck AND cập nhật ô hiện tại = 0

    if room[x][y] == 1:

        print("Action = SUCK")

        room[x][y] = 0


    # RULE 2:
    # IF ô hiện tại = 0
    # AND xét hướng đi
    # THEN predict_state = vị trí kế tiếp

    moves = []

    if room[x][y] == 0:

        # Đi lên
        if x > 0:
            moves.append("U")

        # Đi xuống
        if x < m - 1:
            moves.append("D")

        # Đi trái
        if y > 0:
            moves.append("L")

        # Đi phải
        if y < n - 1:
            moves.append("R")

    print("Possible Move =", moves)


    # Chọn action ngẫu nhiên
    action = random.choice(moves)
    print("Action =", action)

    # Predict state
    predict_state = current_state

    if action == "U":

        predict_state = (x - 1, y)

    elif action == "D":

        predict_state = (x + 1, y)

    elif action == "L":

        predict_state = (x, y - 1)

    elif action == "R":

        predict_state = (x, y + 1)

    print("Predict State =", predict_state)

    # # Bước 4: Hành động 

    # RULE 3:
    # IF predict_state ≠ current_state
    # THEN Move
    # ELSE Stay

    if predict_state != current_state:
        print("Action = MOVE")
        x, y = predict_state

    else:
        print("Action = STAY")

    step += 1
    
    time.sleep(1)

# # RULE 4
# IF tất cả ô = 0
# THEN Stop

print("\nAll clean")
print_room(x, y)
```
