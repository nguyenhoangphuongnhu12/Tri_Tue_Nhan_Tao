```python
#Dùng thư viện from copy import deepcopy để 
#Tạo ra một bản sao y hệt ma trận cũ
#Rồi sửa bản sao đó để thành node con.

from copy import deepcopy
```

```python
#Node class: Lưu trữ cấu trúc dữ liệu Node
class Node:
    def __init__(self, state, parent = None, action = None, 
                 step = 0, depth = 0, name = 'A' ):
        self.state = state
        self.parent = parent
        self.action = action
        self.step = step
        self.depth = depth
        self.name = name
        
    #Hàm biến object thành một chuỗi văn bản (string) để đại diện cho nó
    #Vd in ra REACHED: [A, B, C]    
    def __repr__(self):
        return f"{self.nameham}"
```

```python
#Ma trận 3x3
n = 3

#Hàm in ma trận
def print_state(state):
    for row in state:
        print(row)
    print()
    
#Hàm so sánh hai ma trận
def is_goal(state, goal):
    return state == goal

#Hàm tìm vị trí số 0
def find_zero(state):
    for i in range(n):
        for j in range(n):
            if state[i][j] == 0:
                return i, j
```

```python
#Hàm sinh node con
def generate_children(node):
    children = []
    x, y = find_zero(node.state)

    moves = [
        ("U", -1, 0),
        ("D", 1, 0),
        ("L", 0, -1),
        ("R", 0, 1),
    ]

    #Hàm enumerate() sẽ tự động đính kèm một con số chỉ mục (index) vào trước mỗi phần tử. 
    #index từ 1 đến 4
    #vd: (1, ("U", -1, 0)) | (2, ("D", 1, 0)) ...
    for index, (action, dx, dy) in enumerate(moves, start=1):
        nx = x + dx
        ny = y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = deepcopy(node.state)
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]

            # Tên mới ghép từ Tên_Cha và Số_Thứ_Tự_Con
            child_name = f"{node.name}{index}" 

            child = Node(
                state=new_state,
                parent=node,
                action=action,
                step=node.step + 1,
                depth=node.depth + 1,
                name=child_name,
            )
            children.append(child)

    return children
```

```python
#Hàm in frontier 
def print_frontier(frontier):
    print("\nFRONTIER:")
    if not frontier:
        print("[]")
        return
    
    for node in frontier:
        print(
            f"{node.name}, "
            f"action={node.action}, "
            f"step={node.step}, "
            f"depth={node.depth}"
        )
        print_state(node.state)
```

```python
#Hàm in Reached
def print_reached(reached):
    print("REACHED:", [node.name for node in reached])
```

```python
#Hàm truy vết tìm lời giải
def solution_path(node):
    path = []
    
    while node:
        path.append(node)
        node = node.parent

    path.reverse()
    return path
```

```python
# ===================================================
# Depth Limited Search - APPROACH 2 (Có in ma trận Goal ngay lập tức)
# ===================================================
def depth_limited_search(start, goal, limit):
    # 1. Kiểm tra goal ngay nút gốc trước khi vào vòng lặp
    if is_goal(start.state, goal):
        print(f"GOAL FOUND AT ROOT: {start.name}")
        return start

    frontier = [start]
    reached = [start]  
    
    while frontier:
        # STACK -> LIFO -> pop node cuối
        current = frontier.pop()
        
        print("\n==================================")
        print(f"POP NODE: {current.name}")
        print_state(current.state)

        # Không phải goal -> sinh con
        if current.depth < limit:
            
            children = generate_children(current)
            
            for child in children:
                # Kiểm tra trùng lặp trạng thái
                duplicated = False
                for r in reached:
                    if r.state == child.state:
                        duplicated = True
                        break

                if not duplicated:
                    # =====================================
                    # KIỂM TRA GOAL NGAY KHI SINH NÚT CON (APPROACH 2)
                    # =====================================
                    if is_goal(child.state, goal):
                        print(f"ADD -> {child.name}, parent={current.name}, action={child.action}, step={child.step}, depth={child.depth}")
                        print(f"\nGOAL FOUND AT CHILD: {child.name}")
                        print("MA TRẬN GOAL TÌM THẤY:")  # Thêm dòng này
                        print_state(child.state)         # In ma trận nút con ra luôn tại đây
                        return child

                    frontier.append(child)
                    reached.append(child)  

                    print(
                        f"ADD -> {child.name}, "
                        f"parent={current.name}, "
                        f"action={child.action}, "
                        f"step={child.step}, "
                        f"depth={child.depth}"
                    )

        print_frontier(frontier)
        print_reached(reached)

    return None
```

```python
# Hàm IDDFS
def iddfs(start_state, goal_state, max_depth):
    start_node = Node(
        state=start_state,
        parent=None,
        action=None,
        step=0,
        depth=0,
        name="A"
    )

    for limit in range(max_depth + 1):
        print("\n\n====================================")
        print(f"IDDFS DEPTH = {limit}")

        result = depth_limited_search(start_node, goal_state, limit)
        if result:
            return result

    return None
```

```python
#Hàm chính
start = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

# Chạy thuật toán
result = iddfs(start, goal, max_depth=10)
```

```python

#Hàm IN ĐƯỜNG ĐI LỜI GIẢI
if result:
    print("\n==============================")
    print("SOLUTION PATH")
    print("==============================")

    path = solution_path(result)
    for node in path:
        print(
            f"Node={node.name}, "
            f"action={node.action}, "
            f"step={node.step}, "
            f"depth={node.depth}"
        )
        print_state(node.state)
else:
    print("Không tìm thấy lời giải")
```
